### Data Description

- Customer - Customer ID

- ST - State where customers live

- Gender - Gender of the customer

- Education - Background education of customers

- Customer Lifetime Value - Customer lifetime value(CLV) is the total revenue the client will derive from their entire relationship with a customer. In other words, is the predicted or calculated value of a customer over their entire duration as a policyholder with the insurance company. It is an estimation of the net profit that the insurance company expects to generate from a customer throughout their relationship with the company. Customer Lifetime Value takes into account factors such as the duration of the customer's policy, premium payments, claim history, renewal likelihood, and potential additional services or products the customer may purchase. It helps insurers assess the long-term profitability and value associated with retaining a particular customer.

- Income - Customers income

- Monthly Premium Auto - Amount of money the customer pays on a monthly basis as a premium for their auto insurance coverage. It represents the recurring cost that the insured person must pay to maintain their insurance policy and receive coverage for potential damages, accidents, or other covered events related to their vehicle.

- Number of Open Complaints - Number of complaints the customer opened

- Policy Type - There are three type of policies in car insurance (Corporate Auto, Personal Auto, and Special Auto)

- Vehicle Class - Type of vehicle classes that customers have Two-Door Car, Four-Door Car SUV, Luxury SUV, Sports Car, and Luxury Car

- Total Claim Amount - the sum of all claims made by the customer. It represents the total monetary value of all approved claims for incidents such as accidents, theft, vandalism, or other covered events.

## Solution setup

We start by importing pandas and loading the CSV into a DataFrame named `df`.

- `pd` is the conventional short name for pandas.
- `pd.read_csv(...)` reads a comma-separated file into a table.
- `head()` is a quick visual check that the file loaded correctly.


```python
import pandas as pd

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv"
df = pd.read_csv(url)
df.head()
```

### 1. Dimensions and basic structure

`shape` returns `(number of rows, number of columns)`. `info()` reports column names, non-null counts, and inferred data types. The non-null counts also reveal missing values.


```python
rows, columns = df.shape
print(f"Rows: {rows:,}")
print(f"Columns: {columns}")
print()
df.info()
```

### 2. Data types and suggested corrections

`dtypes` shows pandas' current interpretation. Text usually appears as `object`. Several columns need attention:

- **Customer Lifetime Value** is text because its values contain `%`; remove the symbol and convert to numeric.
- **Number of Open Complaints** looks like `1/2/00`; the middle component is the complaint count, so extract it as an integer.
- **Customer**, **ST**, **GENDER**, **Education**, **Policy Type**, and **Vehicle Class** are labels and can use the `category` dtype.
- The remaining measure columns should be numeric. `errors="coerce"` safely changes invalid values to `NaN`, making data problems visible rather than crashing.


```python
print(df.dtypes)

# Work on a copy so the original raw data remains available for comparison.
clean_df = df.copy()

clean_df["Customer Lifetime Value"] = pd.to_numeric(
    clean_df["Customer Lifetime Value"].str.replace("%", "", regex=False),
    errors="coerce"
)

clean_df["Number of Open Complaints"] = pd.to_numeric(
    clean_df["Number of Open Complaints"].str.split("/").str[1],
    errors="coerce"
).astype("Int64")

numeric_columns = ["Income", "Monthly Premium Auto", "Total Claim Amount"]
clean_df[numeric_columns] = clean_df[numeric_columns].apply(
    pd.to_numeric, errors="coerce"
)

categorical_columns = [
    "Customer", "ST", "GENDER", "Education", "Policy Type", "Vehicle Class"
]
clean_df[categorical_columns] = clean_df[categorical_columns].astype("category")

clean_df.dtypes
```

### 3. Unique values, categorical variables, and numeric ranges

`nunique(dropna=False)` counts distinct values including missing values. Columns with a small, repeated set of labels are categorical; **Customer** is an identifier, not a true category for analysis, even though it is text.

For categorical variables we list their distinct labels. For numeric variables we calculate minimum and maximum. This also helps spot inconsistent labels (for example state abbreviations versus full names) and suspicious outliers.


```python
unique_counts = clean_df.nunique(dropna=False).sort_values()
display(unique_counts.rename("number_of_unique_values"))

analysis_categories = ["ST", "GENDER", "Education", "Policy Type", "Vehicle Class"]
for column in analysis_categories:
    values = clean_df[column].astype("object").dropna().unique()
    print(f"{column}: {sorted(map(str, values))}")

numeric_cols = clean_df.select_dtypes(include="number").columns
numeric_ranges = clean_df[numeric_cols].agg(["min", "max"]).T
display(numeric_ranges)
```

**Interpretation:** A high unique count in `Customer` is expected because it is an ID. `ST`, `GENDER`, and `Education` should be standardized before production analysis because variants such as abbreviations or alternate spellings may represent the same group. Very large values in `Monthly Premium Auto` should be checked as possible data-entry errors.


### 4. Numerical summary statistics

`describe()` provides count, mean, standard deviation, minimum, quartiles, and maximum. We add median and mode explicitly. Compare mean with median: a large difference suggests skewness or outliers. The interquartile range is `75% - 25%` and describes the middle half of observations.


```python
numerical_summary = clean_df[numeric_cols].describe().T
numerical_summary["median"] = clean_df[numeric_cols].median()
numerical_summary["mode"] = clean_df[numeric_cols].mode().iloc[0]
numerical_summary["IQR"] = numerical_summary["75%"] - numerical_summary["25%"]
numerical_summary
```

**Interpretation:** Use `count` to notice missing numeric values. When the maximum is far above the 75th percentile, the distribution has a long upper tail or an outlier. Income values of zero may be legitimate, missing information encoded as zero, or customers without reported income; the business meaning must be confirmed.


### 5. Categorical summary statistics

For categorical columns, `describe()` reports the non-null count, number of unique labels, most frequent label (`top`), and its frequency (`freq`). Missing counts are added because they are important for data quality.


```python
categorical_summary = clean_df[analysis_categories].describe().T
categorical_summary["missing"] = clean_df[analysis_categories].isna().sum()
categorical_summary
```

**Interpretation:** `top` identifies the dominant group in each column, while `freq / count` gives its approximate share. A very dominant label can create class imbalance. Missing gender values and inconsistent state or education labels should be cleaned before group comparisons.

### Exercise 3

The sales team wants to know if customers with Personal Auto have a lower income than those with Corporate Auto. How does the average income compare between the two policy types?

- Use *loc* to create two dataframes: one containing only Personal Auto policies and one containing only Corporate Auto policies.
- Calculate the average income for each policy.
- Print the results.

### Solution

`.loc[condition]` filters rows where the condition is true. We create the two requested DataFrames and compute the arithmetic mean of `Income`. Pandas ignores missing values in `mean()` by default, but zeros remain included because they are actual stored values.


```python
personal_auto = df.loc[df["Policy Type"] == "Personal Auto"].copy()
corporate_auto = df.loc[df["Policy Type"] == "Corporate Auto"].copy()

personal_mean_income = personal_auto["Income"].mean()
corporate_mean_income = corporate_auto["Income"].mean()

print(f"Personal Auto average income:  ${personal_mean_income:,.2f}")
print(f"Corporate Auto average income: ${corporate_mean_income:,.2f}")

if personal_mean_income < corporate_mean_income:
    print("Conclusion: Personal Auto customers have a lower average income.")
else:
    print("Conclusion: Personal Auto customers do not have a lower average income.")
```

**Important analytical note:** If zero means “income not reported” instead of true zero income, replace zero with `pd.NA` before calculating the averages and state that assumption. Otherwise, zero values can pull both averages downward.

A percentile is a measure that tells us what percentage of values in a dataset are below a certain value. For example, the 75th percentile represents the value below which 75% of the data falls. Similarly, the 25th percentile represents the value below which 25% of the data falls. When we talk about the top 25%, we are referring to the values that fall above the 75th percentile, which represent the top quarter of the data. On the other hand, when we talk about the bottom 25%, we are referring to the values that fall below the 25th percentile, which represent the bottom quarter of the data. ### Solution

First we review the distribution. `quantile(0.75)` returns the 75th percentile: 75% of observed claim amounts are at or below this threshold. A Boolean condition with `>` then selects customers strictly above it, which matches “greater than the 75th percentile.”


```python
claim_summary = df["Total Claim Amount"].describe()
display(claim_summary)

claim_75th_percentile = df["Total Claim Amount"].quantile(0.75)
high_claim_customers = df.loc[
    df["Total Claim Amount"] > claim_75th_percentile
].copy()

print(f"75th-percentile threshold: ${claim_75th_percentile:,.2f}")
print(f"Customers above threshold: {len(high_claim_customers):,}")
display(high_claim_customers.head())
```

Now `describe(include="all")` summarizes both numeric and categorical fields for the high-claim group. The second table focuses on claim amounts, while the policy counts show whether a particular policy type is common among these customers.


```python
display(high_claim_customers.describe(include="all").T)

display(
    high_claim_customers["Total Claim Amount"]
    .describe()
    .rename("high_claim_summary")
)

display(
    high_claim_customers["Policy Type"]
    .value_counts()
    .rename("high_claim_customer_count")
)
```

**Interpretation:** Every selected claim is above the overall 75th-percentile threshold. Compare this subgroup's mean and median claim with the full-data statistics to measure how much larger these claims are. Policy-type counts are useful descriptively, but comparing raw counts alone can be misleading if the policy types have very different numbers of customers overall; rates by policy type would be the next analysis.
